```python
import pandas as pd

train_df= pd.read_csv("../final_data/train/dico_nli_train_track1_participant_labeled.csv")

dev_df=pd.read_csv(  "../final_data/dev/dico_nli_dev_track1_participant_labeled.csv")

print(train_df.shape)
print(dev_df.shape)
```

```python
dev_df["exact_match"]=(
    dev_df["text1"].str.lower().str.strip() == dev_df["text2"].str.lower().str.strip()
)

dev_df[["text1", "text2", "label", "exact_match"]].head(10)
```

```python
dev_df[dev_df["exact_match"]]["label"].value_counts()

dev_df["exact_match"].value_counts()
```

```python
dev_df["text2_in_text1"]=dev_df.apply( lambda row: row["text2"].lower().strip() in row["text1"].lower().strip(), axis=1)

dev_df["text1_in_text2"]=dev_df.apply( lambda row: row["text1"].lower().strip() in row["text2"].lower().strip(), axis=1)

print(dev_df["text1_in_text2"].value_counts())
print(dev_df["text2_in_text1"].value_counts())
```

```python
print("text2 in text1")
print(
    dev_df[dev_df["text2_in_text1"]]["label"].value_counts()
)
print("\ntext1 in text2")
print(
    dev_df[dev_df["text1_in_text2"]]["label"].value_counts()
)
```

```python
train_df["text2_in_text1"]=train_df.apply( lambda row: row["text2"].lower().strip() in row["text1"].lower().strip(), axis=1)

train_df["text1_in_text2"]=train_df.apply( lambda row: row["text1"].lower().strip() in row["text2"].lower().strip(), axis=1)

print(train_df["text1_in_text2"].value_counts())
print(train_df["text2_in_text1"].value_counts())


print("text2 in text1")
print(
    train_df[train_df["text2_in_text1"]]["label"].value_counts()
)
print("\ntext1 in text2")
print(
    train_df[train_df["text1_in_text2"]]["label"].value_counts()
)
```

```python
neither=train_df[
    (~train_df["text2_in_text1"] & ~train_df["text1_in_text2"])

]
neither["label"].value_counts()
```

```python
def word_subset_features(row):
    text1_words = set(row["text1"].lower().strip().split())
    text2_words = set(row["text2"].lower().strip().split())

    return pd.Series({
        "text2_words_in_text1": text2_words.issubset(text1_words),
        "text1_words_in_text2": text1_words.issubset(text2_words)
    })
```

```python
train_df[
    ["text2_words_in_text1", "text1_words_in_text2"]
] = train_df.apply(
    word_subset_features,
    axis=1
)

dev_df[
    ["text2_words_in_text1", "text1_words_in_text2"]
] = dev_df.apply(
    word_subset_features,
    axis=1
)
```

```python
print("Train - text2 words in text1:")
print(train_df["text2_words_in_text1"].value_counts())

print("\nTrain - text1 words in text2:")
print(train_df["text1_words_in_text2"].value_counts())

print("\nDev - text2 words in text1:")
print(dev_df["text2_words_in_text1"].value_counts())

print("\nDev - text1 words in text2:")
print(dev_df["text1_words_in_text2"].value_counts())
```

```python
print("text2 words in text1:")
print(
    train_df[train_df["text2_words_in_text1"]]["label"].value_counts()
)

print("\ntext1 words in text2:")
print(
    train_df[train_df["text1_words_in_text2"]]["label"].value_counts()
)
```

```python
both_true=train_df[
    (train_df["text2_words_in_text1"]) & (train_df["text1_words_in_text2"])
]

print("number of examples:",len(both_true))
print(both_true["label"].value_counts())

both_true[["text1","text2","label"]].head(10)
```

```python
def lexical_predict(row):
    text1_words=set(row["text1"].lower().strip().split())
    text2_words=set(row["text2"].lower().strip().split())

    text2_in_text1= text2_words.issubset(text1_words)
    text1_in_text2= text1_words.issubset(text2_words)

    if text2_in_text1 and text1_in_text2:
        return "EQUIVALENCE"
    elif text2_in_text1: 
        return "FORWARD_ENTAILMENT"
    elif text1_in_text2:
        return "BACKWARD_ENTAILMENT"
    else:
        return "EQUIVALENCE"
```

```python
dev_df["prediction"]=dev_df.apply(lexical_predict, axis=1)

dev_df[["text1","text2","label","prediction"]].head(10)
```

```python
lexical_submission=dev_df[["instance_id"]].copy()
lexical_submission["label"]=dev_df["prediction"]

lexical_submission.head()
```

```python
lexical_submission.to_csv("../submissions/lexical_baseline_dev.csv", index=False)
```
